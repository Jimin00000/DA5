## 2-4. 새로운 피처 생성
결측치는 없음을 1-1 `info()`로 확인했으므로 제외   
이상치는 사기거래 예측에 중요한 정보일 수 있으므로 처리 제외  

- 거래 금액 : 로그 변환 및 고액 거래 여부 확인 컬럼    
- 거래 발생 시간대 : 요일 및 시간대 확인 컬럼  
- 가맹점 & 고객 위치 거리 확인  
<br>
- 고객별 평균 거래금액 대비 현재 금액 차이
- 카테고리별 평균 금액
- 고객별 하루 거래 횟수
- 나이 계산 (dob 활용)
- 도시 인구 대비 거래 패턴

```python
# 거래 금액 : 로그 변환 (이상치 제거 없이 분포 안정화)
cc_df["log_amt"] = np.log1p(cc_df["amt"])

# 고액 거래 여부(전역 95% 기준)
amt_95 = cc_df["amt"].quantile(0.95)
cc_df["high_amount_flag"] = (cc_df["amt"] >= amt_95).astype(int)

# 상위 고액 거래 10건 확인
cc_df.sort_values(by="amt", ascending=False).head(10)
```

```python
# 거래 발생 시간대 : 주말 여부
cc_df['is_weekend'] = cc_df['weekday'].isin([5,6]).astype(int)

# 거래 발생 시간대 : 심야 거래 여부 (00~05시)
cc_df['is_night'] = cc_df['hour'].between(0, 5).astype(int)
```

```python
# 가맹점 & 고객 위치 거리 (Haversine)
def haversine_km(lat1, lon1, lat2, lon2):
    """
    위경도(도 단위) -> 거리(km)
    """
    R = 6371.0  # 지구 반지름 (km)
    lat1 = np.radians(lat1); lon1 = np.radians(lon1)
    lat2 = np.radians(lat2); lon2 = np.radians(lon2)
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1)*np.cos(lat2)*np.sin(dlon/2)**2
    c = 2*np.arcsin(np.sqrt(a))
    return R * c

cc_df["distance_km"] = haversine_km(
    cc_df["lat"], cc_df["long"], cc_df["merch_lat"], cc_df["merch_long"]
)

# 장거리 플래그(전역 95% 기준) - 거리도 이상치가 "신호"일 수 있어 플래그만 생성
dist_95 = cc_df["distance_km"].quantile(0.95)
cc_df["far_distance_flag"] = (cc_df["distance_km"] >= dist_95).astype(int)
```

```python
# 고객별 평균 거래금액 대비 현재 금액 차이

# 평균 거래 금액
cc_df["cust_mean_amt"] = cc_df.groupby("cc_num")["amt"].transform("mean")

# 평소보다 얼마나 많이/적게 썼는가? (현재 거래금액 - 고객 평균금액)
cc_df["amt_minus_cust_mean"] = cc_df["amt"] - cc_df["cust_mean_amt"]

# 평균 대비 몇 배인가? (현재 금액 ÷ 고객 평균)
cc_df["amt_over_cust_mean"] = cc_df["amt"] / cc_df["cust_mean_amt"]
```

```python
# 카테고리별 평균 금액 및 차이
cc_df["cat_mean_amt"] = cc_df.groupby("category")["amt"].transform("mean")
cc_df["amt_minus_cat_mean"] = cc_df["amt"] - cc_df["cat_mean_amt"]
cc_df["amt_over_cat_mean"] = cc_df["amt"] / cc_df["cat_mean_amt"]
```

```python
# 고객별 하루 거래 횟수
cc_df["cust_daily_txn_count"] = cc_df.groupby(["cc_num", "tx_date"])["amt"].transform("size")
```

```python
# 거래 시점 나이
cc_df["dob"] = pd.to_datetime(cc_df["dob"], errors="coerce")
cc_df["age"] = ((cc_df["trans_date_trans_time"] - cc_df["dob"]).dt.days / 365.25).astype("float")

# 혹시 결측/변환 실패가 있으면 age에 NaN이 생길 수 있어 처리
cc_df["age"] = cc_df["age"].fillna(cc_df["age"].median())
```

```python
# 도시 인구 대비 거래 패턴
cc_df["log_city_pop"] = np.log1p(cc_df["city_pop"])
cc_df["amt_per_city_pop"] = cc_df["amt"] / (cc_df["city_pop"] + 1)  # 0 나눗셈 방지
```

# 4️⃣ 모델 학습 - Logistic Regression

Target이 bool타입이므로, 이진 분류에 최적화된 확률 기반 모델인 로지스틱 회귀로 학습  

사기일 확률이 0.25%이므로, 매우 큰 클래스 불균형 상태임  
그래서 class_weight="balanced" (가중치 조정) 옵션 추가하여 모델링 진행

```python
# 기본 모델 학습 (Logistic Regression)
from sklearn.linear_model import LogisticRegression

logit = LogisticRegression(max_iter=1000, class_weight="balanced")
logit.fit(X_train, y_train)
```

```python
from sklearn.metrics import (
    recall_score, f1_score,
    roc_auc_score, average_precision_score
)

# 클래스 예측
y_pred = logit.predict(X_test)

# 사기(1)일 확률
y_proba = logit.predict_proba(X_test)[:, 1]

# 지표 계산
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_proba)
pr_auc = average_precision_score(y_test, y_proba)

print("Recall:", recall)
print("F1 Score:", f1)
print("ROC-AUC:", roc_auc)
print("PR-AUC:", pr_auc)
```

- Recall = 0.64  
실제 사기 거래 중 64%를 잡음  

- F1 = 0.049  
Precision이 매우 낮다는 뜻
